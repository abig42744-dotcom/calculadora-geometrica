```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
print("----------------------------------")
print("CALCULADORA DE GEOMTERIA ANALITICA")
print("----------------------------------")
```

```python
#BLOQUE 2.ENTRADA DE DATOS

x1 = float(input("Ingresa X del punto A: "))
y1 = float(input("Ingresa Y del punto A: "))

x2 = float(input("Ingresa X del punto A: "))
y2 = float(input("Ingresa Y del punto A: "))

print("Punto A:", (x1, y1))
print("Punto B:", (x2, y2))
```

```python
# BLOQUE 3. VALIDACION DE PUNTOS

if x1 == x2 and y1 == y2:
  print("ERROR: Los puntos A y B son iguales. ")
  print("No se puede determinar una recta unica. ")
  raise SystemExit("Ejecuta nuavamente desde el bloque 2.")

print("Validacion correcta: los puntos son diferentes. ")
```

```python
#BLOQUE 4. DIFERENCIAS DE COORDENADAS
# Crear una variable e igualarla
dx = x2 - x1
dy= y2 - y1

print ("Diferencia en X:" , dx)
print ("Diferencia en Y:" , dy)
```

```python
#BLOQUE 5. CALCULO DISTANCIAS

distancia = math.sqrt(dx**2 + dy**2)
print("Distancia entre A y B:" ,round (distancia,2))
```

```python
#BLOQUE 6. PUNTO MEDIO
mx = (x1+x2)/2
my= (y1+y2)/2

print("Punto medio en X:" ,mx)
print("Punto medio en Y:" ,my)

print("El punto medio entre A y B es:" ,(mx,my))
```

```python
#BLOQUE 7. VALIDAR PENDIENTE

if dx == 0:
 print("La recta es vertical.")
 print("Su pendiente no esta definida.")

else:
 print("La pendiente se puede calcular.")
```

```python
#BLOQUE 8.

if dx != 0:
  pendiente = dy / dx

  if pendiente > 0:
    tipo_recta = "Creciente"
  elif pendiente < 0:
    tipo_recta = "Decreciente"
  else:
    tipo_recta = "Horizontal"

else:
  pendiente = None
  tipo_recta = "Vertical"

print("Pendiente:" , pendiente)
print("Tipo de recta:" ,tipo_recta)
```

```python
#BLOQUE 9. ECUACION DE LA RECTA

if pendiente is not None:
  b = y1 - pendiente * x1

  ECUACION = f"x = {pendiente:.2f} x + ({b:.2f})"
else:
  b = None
  ECUACION = f"x = {x1:.2f}"

print("Ecuacion de la recta:")
print(ECUACION)
```

```python
#BLOQUE 10. MOSTRAR RESULTADOS

print("\n ========== RESULTADOS ========== ")

print("Punto A:" , (x1, y1))
print("Punto B:" ,(x2,y2))

print("Distancia:" , round(distancia,2))
print("Punto medio:" ,(mx,my))

print("Tipo de recta:",tipo_recta)
print("Pendiente:" , pendiente)
print("Ecuacion" , ECUACION)
```

```python
#BLOQUE 11. GRAFICAR PUNTOS

plt.figure(figsize=(7,7))

plt.scatter(x1 , y1, color="blue" , s=90, label="Punto A")
plt.scatter(x2 , y2, color="green" , s=90, label="Punto B")
plt.scatter(mx , my, color="red" , s=90, label="Punto medio")

plt.annotate("A", (x1,y1), xytext=(6,6),textcoords="offset points")
plt.annotate("B", (x2,y2), xytext=(6,6),textcoords="offset points")
plt.annotate("M", (mx,my), xytext=(6,6),textcoords="offset points")

plt.axhline(0, color="black")
plt.axvline(0, color="black")

plt.grid(True)
plt.axis("equal")
plt.legend()
plt.title("Representacion de dos puntos y su punto medio")

plt.show()
```

```python
def calcular_vector (x1,y1 ,x2,y2):
    vx = x2 - x1 
    vy = y2 - y1

    magnitud =math.sqrt(vx**2 + vy**2)

    if magnitud == 0:
        angulo = None
        unitario = None
    else:
        angulo = math.degrees(math.atan2(vy ,vx)) % 360

        unitario = (
            vx / magnitud ,
            vy / magnitud
        )

    return ( vx , vy ),magnitud , angulo, unitario

#MOSTRAR RESULTADOS:
vector , magnitud , angulo ,unitario = calcular_vector (2 ,1 ,8 ,9 )
print ("Vector AB:" , vector)
print ("Magnitud " , magnitud)
print("Direccion" , angulo)
print ("Vector unitario " , unitario)
```
